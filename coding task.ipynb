```python

#Q1: FizzBuzz Challenge
print("=" * 60)
print("Q1: FizzBuzz Challenge (1 to 50)")
print("=" * 60)

for i in range(1, 51):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)





#"Q2: Leap Year Logic
print("\n" + "=" * 60)
print("Q2: Leap Year Logic")
print("=" * 60)

def is_leap_year(year):
    """Return True if 'year' is a leap year, else False."""
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

for test_year in [1900, 2000, 2024, 2023]:
    print(f"{test_year} -> {is_leap_year(test_year)}")





#Q3: Identity vs Equality
print("\n" + "=" * 60)
print("Q3: Identity vs Equality")
print("=" * 60)

L1 = [1, 2, 3]
L2 = [1, 2, 3]

print("L1 == L2:", L1 == L2)   # True: compares VALUES (element by element)
print("L1 is L2:", L1 is L2)   # False: compares IDENTITY (same object in memory)
print("id(L1):", id(L1), " id(L2):", id(L2))





#Q4: Bitwise Swap (Advanced)
print("\n" + "=" * 60)
print("Q4: Bitwise Swap (Advanced)")
print("=" * 60)

a, b = 5, 9
print(f"Before swap: a = {a}, b = {b}")

a = a ^ b
b = a ^ b
a = a ^ b

print(f"After swap:  a = {a}, b = {b}")



#Q5: Prime Number Finder
print("\n" + "=" * 60)
print("Q5: Prime Number Finder (first 10 primes)")
print("=" * 60)

def is_prime(n):
    if n < 2:
        return False
    i = 2
    while i * i <= n:
        if n % i == 0:
            return False
        i += 1
    return True

count = 0
num = 2
primes_found = []
while count < 10:
    if is_prime(num):
        primes_found.append(num)
        count += 1
    num += 1

print("First 10 primes:", primes_found)



#Q6: The Loop Breaker (1 to 20)
print("\n" + "=" * 60)
print("Q6: The Loop Breaker (1 to 20)")
print("=" * 60)

for i in range(1, 21):
    if i % 4 == 0:
        pass  # explicitly do nothing for multiples of 4
    if i == 13:
        continue  # skip the rest of this iteration for 13
    if i == 18:
        break  # stop the loop entirely at 18
    print(i)



#Q7: Nested Conditionals
print("\n" + "=" * 60)
print("Q7: Nested Conditionals (Grade Calculator)")
print("=" * 60)

def get_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    else:
        return "Fail"

for s in [95, 82, 71, 50]:
    print(f"Score {s} -> Grade {get_grade(s)}")




#Q8: Reverse an Integer
print("\n" + "=" * 60)
print("Q8: Reverse an Integer")
print("=" * 60)

def reverse_integer(n):
    negative = n < 0
    n = abs(n)
    reversed_num = 0
    while n > 0:
        digit = n % 10
        reversed_num = reversed_num * 10 + digit
        n = n // 10
    return -reversed_num if negative else reversed_num

for num in [12345, 900, -6789]:
    print(f"{num} -> {reverse_integer(num)}")




#Q9: Runner-Up Score (no sort/sorted)
print("\n" + "=" * 60)
print("Q9: Runner-Up Score (no sort/sorted)")
print("=" * 60)

def runner_up_score(scores):
    highest = second_highest = float("-inf")
    for s in scores:
        if s > highest:
            second_highest = highest
            highest = s
        elif highest > s > second_highest:
            second_highest = s
    return second_highest

test_scores = [10, 20, 20, 30, 5, 30, 25]
print("Scores:", test_scores)
print("Runner-up score:", runner_up_score(test_scores))





#Q10: Duplicate Removal (one-liner

print("\n" + "=" * 60)
print("Q10: Duplicate Removal (one-liner)")
print("=" * 60)

original_list = [1, 2, 2, 3, 4, 4, 5, 1]
unique_values = list(set(original_list))
print("Original:", original_list)
print("Unique:  ", unique_values)



#Q11: List Intersection (using 'in')

print("\n" + "=" * 60)
print("Q11: List Intersection (using 'in')")
print("=" * 60)

list_a = [1, 2, 3, 4, 5]
list_b = [3, 4, 5, 6, 7]

common_elements = []
for item in list_a:
    if item in list_b:
        common_elements.append(item)

print("List A:", list_a)
print("List B:", list_b)
print("Common elements:", common_elements)




#Q12: Tuple Immutability Test
print("\n" + "=" * 60)
print("Q12: Tuple Immutability Test")
print("=" * 60)

my_tuple = (1, 2, 3)
try:
    my_tuple[1] = 99
except TypeError as e:
    print(f"Error caught: {e}")
    print("Tuples are immutable — their elements cannot be reassigned.")




#Q13: Character Frequency Counter
print("\n" + "=" * 60)
print("Q13: Character Frequency Counter")
print("=" * 60)

def char_frequency(text):
    freq = {}
    for ch in text:
        freq[ch] = freq.get(ch, 0) + 1
    return freq

sample_text = "hello world"
print(f"Input: '{sample_text}'")
print("Frequency:", char_frequency(sample_text))





#Q14: List Comprehension (squares of evens 1-20)
print("\n" + "=" * 60)
print("Q14: List Comprehension (squares of evens 1-20)")
print("=" * 60)

squares_of_evens = [x**2 for x in range(1, 21) if x % 2 == 0]
print("Squares of even numbers (1-20):", squares_of_evens)



#Q15: Dictionary Comprehension (values > 2)

print("\n" + "=" * 60)
print("Q15: Dictionary Comprehension (values > 2)")
print("=" * 60)

original_dict = {'a': 1, 'b': 2, 'c': 3, 'd': 4}
filtered_dict = {k: v for k, v in original_dict.items() if v > 2}
print("Original dict:", original_dict)
print("Filtered dict (values > 2):", filtered_dict)
```
